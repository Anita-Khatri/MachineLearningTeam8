```python
import numpy as np
import pandas as pd
```

```python
#1.Load the dataset
#We first load the training file (adult.data).
#The Adult dataset uses commas as separators and may have leading spaces after commas, so we use skipinitialspace=True.

df=pd.read_csv("adult.data")
columns = [
    "age", "workclass", "fnlwgt", "education", "education_num",
    "marital_status", "occupation", "relationship", "race", "sex",
    "capital_gain", "capital_loss", "hours_per_week", "native_country", "income"
]
adult_df=pd.read_csv("adult.data",
    names=columns,
    sep=",",
    skipinitialspace=True
)
# shape returns a tuple(number_of_rows, number_of_columns)
adult_df.shape
```

```python
#2.Taking a quick look at the first few rows of the dataset using the head() function.
adult_df.head()
```

```python
#3.Taking a quick look at the first few rows of the dataset using the sample()(which shows random rows) function.
# (random_state=42)-It fixes the randomness.Every time you get the same 5 rows.
adult_df.sample(5, random_state=42)
```

```python
#4.Using [info()(It gives 1.number of rows and columns,2.data types (numeric vs object),3.non-null counts)] to get an
#overview of the dataset’s columns and their data types.
adult_df.info()
```

```python
#5.Using [isnull()( checks for NaN/None values.)] to find and understand missing values in the data.
adult_df.isnull().sum()
```
